## Setup and Imports

```python
# Import required libraries
import sys
import os

# Add src directory to path
sys.path.append(os.path.join(os.getcwd(), '..', 'src'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Import custom modules
from data_loader import DataLoader
from data_cleaner import DataCleaner
from database import DebtDatabase
from visualizer import DebtVisualizer

# Configure display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.float_format', '{:.2f}'.format)

# Set visualization style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

print("✓ All libraries imported successfully")
```

## 1. Data Loading and Initial Exploration

```python
# Check if dataset is ready
loader = DataLoader(data_dir="../data/raw")
readiness = loader.check_dataset_readiness()

if not readiness["ready"]:
    print("⚠️  Dataset Not Found!")
    print(f"Message: {readiness['message']}")
    print("\nInstructions:")
    for instruction in readiness["instructions"]:
        print(f"  - {instruction}")
else:
    print("✓ All dataset files are present")
    print("\nFiles found:")
    for filename, status in readiness['files_status'].items():
        print(f"  ✓ {filename}")
```

```python
# Load the primary data file (IDSData.csv)
success, message = loader.load_primary_data()
print(message)

if success:
    df = loader.get_primary_data()
    print("\nPrimary data loaded successfully!")
    print(f"Shape: {df.shape}")
    print(f"\nColumns: {list(df.columns)}")
```

```python
# Display first few rows
print("First 5 rows of the dataset:")
df.head()
```

```python
# Get basic information about the primary data
info = loader.get_basic_info('data')
print("Primary Data Information:")
print(f"File: {info['file']}")
print(f"Total Rows: {info['rows']}")
print(f"Total Columns: {info['columns']}")
print(f"\nColumn Names: {info['column_names']}")
print(f"\nData Types:")
for col, dtype in info['data_types'].items():
    print(f"  {col}: {dtype}")
print(f"\nMemory Usage: {info['memory_usage_mb']:.2f} MB")
```

```python
# Check for missing values
print("Missing values per column:")
missing = df.isnull().sum()
missing[missing > 0]
```

## 2. Data Cleaning

```python
# Initialize data cleaner
cleaner = DataCleaner(df)

# Remove duplicates
cleaner.remove_duplicates()

# Handle missing values
cleaner.handle_missing_values(strategy="drop")

# Get cleaned data
df_clean = cleaner.get_cleaned_data()

# Display cleaning summary
summary = cleaner.get_cleaning_summary()
print("Cleaning Summary:")
print(f"Original shape: {summary['original_shape']}")
print(f"Cleaned shape: {summary['cleaned_shape']}")
print(f"Rows removed: {summary['rows_removed']}")
print("\nOperations performed:")
for op in summary['cleaning_operations']:
    print(f"  - {op}")
```
